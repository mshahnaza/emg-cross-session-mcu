```python
import os
from emg.data import extract_subject, save_features

SUBJECTS = range(1, 11)
TMP = '/content/db6/'
URL = 'https://ninapro.hevs.ch/files/DB6_Preproc/'

for s in SUBJECTS:
    out = f'/content/drive/MyDrive/ninapro/features_S{s}.npz'
    if os.path.exists(out):
        print(f'S{s}: features already saved')
        continue
    !rm -rf {TMP} && mkdir -p {TMP}
    for part in ['a', 'b']:
        !wget -q -c {URL}DB6_s{s}_{part}.zip -P {TMP}
        !unzip -q -o -j {TMP}DB6_s{s}_{part}.zip "*.mat" -d {TMP}
    features = extract_subject(TMP, s)
    save_features(features, out)
    print(f'S{s}: saved')

!rm -rf {TMP}
```
